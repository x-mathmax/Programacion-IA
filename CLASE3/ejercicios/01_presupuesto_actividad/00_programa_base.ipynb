# Programa 1 · Presupuesto de una actividad

Este programa representa una actividad mediante una clase. El objeto guarda sus datos y sus métodos calculan el costo total, el costo por persona y el estado del presupuesto.

**Estructuras que aparecen:** variables tipadas, listas, bucles, funciones, condicionales, una clase, un objeto y métodos.

Ejecutá las celdas en orden. El programa comienza en `MODO_DEMO` para que la primera ejecución no solicite datos.

## 1. Clase `Actividad`

La clase funciona como un modelo. Cada objeto creado a partir de ella tendrá un nombre, un presupuesto, una cantidad estimada de asistentes y su propia lista de rubros.

```python
class Actividad:
    def __init__(
        self,
        nombre: str,
        presupuesto: float,
        asistentes: int,
    ) -> None:
        # Estos atributos guardan los datos propios de cada actividad.
        self.nombre: str = nombre
        self.presupuesto: float = presupuesto
        self.asistentes: int = asistentes
        self.rubros: list[tuple[str, float]] = []

    def agregar_rubro(self, nombre: str, costo: float) -> None:
        # El nuevo rubro se agrega a la lista del objeto.
        self.rubros.append((nombre, costo))

    def calcular_total(self) -> float:
        total: float = 0.0
        for _, costo in self.rubros:
            total += costo
        return total

    def calcular_costo_por_persona(self) -> float:
        if self.asistentes <= 0:
            return 0.0
        return self.calcular_total() / self.asistentes

    def evaluar_presupuesto(self) -> str:
        diferencia: float = self.presupuesto - self.calcular_total()

        if diferencia >= 0:
            return f"Viable: quedan ${diferencia:,.2f} disponibles"

        return f"No viable: faltan ${abs(diferencia):,.2f}"
```

## 2. Creación del objeto

Primero definimos los datos. Después creamos `actividad`, un objeto concreto de la clase `Actividad`.

```python
MODO_DEMO: bool = True

if MODO_DEMO:
    nombre_actividad: str = "Muestra de proyectos"
    presupuesto_disponible: float = 350_000.0
    asistentes_estimados: int = 80
else:
    nombre_actividad = input("Nombre de la actividad: ")
    presupuesto_disponible = float(input("Presupuesto disponible: "))
    asistentes_estimados = int(input("Asistentes estimados: "))

actividad: Actividad = Actividad(
    nombre=nombre_actividad,
    presupuesto=presupuesto_disponible,
    asistentes=asistentes_estimados,
)

print(f"Objeto creado: {actividad.nombre}")
```

## 3. Incorporación de rubros

Usamos el método `agregar_rubro()` para cargar los costos dentro del objeto.

```python
actividad.agregar_rubro("Espacio", 120_000.0)
actividad.agregar_rubro("Técnica", 95_000.0)
actividad.agregar_rubro("Comunicación", 38_000.0)
actividad.agregar_rubro("Catering", 72_000.0)

print(actividad.rubros)
```

## 4. Ejecución e informe

```python
total_actividad: float = actividad.calcular_total()
valor_por_persona: float = actividad.calcular_costo_por_persona()
resultado: str = actividad.evaluar_presupuesto()

print(f"ACTIVIDAD: {actividad.nombre}")
print("-" * 48)

for rubro, costo in actividad.rubros:
    print(f"{rubro:<20} ${costo:>12,.2f}")

print("-" * 48)
print(f"TOTAL:               ${total_actividad:>12,.2f}")
print(f"POR PERSONA:         ${valor_por_persona:>12,.2f}")
print(f"PRESUPUESTO:         ${actividad.presupuesto:>12,.2f}")
print(resultado)
```
