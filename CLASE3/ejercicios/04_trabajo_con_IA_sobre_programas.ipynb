## 5. Tercer prompt: pedir una modificación acotada

Cuando el plan sea claro, pedimos solamente la parte necesaria:

> Escribí únicamente el método o la función necesaria para incorporar esa funcionalidad. Conservá los nombres, los tipos y el estilo del programa original. Indicá exactamente dónde debo colocar el código. No reescribas el programa completo.

Copiá primero el código sugerido en una celda nueva. No reemplaces inmediatamente la versión que funciona.


```python
# Pegá aquí la modificación sugerida por la IA.
# Esta celda funciona como registro y puede permanecer comentada.
def __init__(self, nombre, presupuesto, asistentes, porcentaje_imprevistos=0.0):
    ...
    self.porcentaje_imprevistos: float = porcentaje_imprevistos

def calcular_subtotal(self) -> float:
    total = 0.0
    for _, costo in self.rubros:
        total += costo
    return total

def calcular_imprevistos(self) -> float:
    return self.calcular_subtotal() * (self.porcentaje_imprevistos / 100)

def calcular_total(self) -> float:
    return self.calcular_subtotal() + self.calcular_imprevistos()
```
